```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("C:/Users/PUNITHA/Downloads/archive (2)/fake_news_dataset.csv")
df
```

# HISTOGRAM

```python
df['title_length'] = df['title'].astype(str).apply(len)

plt.figure(figsize=(8,3))
plt.hist(df['title_length'], bins=30, edgecolor='black')

plt.title("Histogram of News Title Length")
plt.xlabel("Title Length")
plt.ylabel("Frequency")
plt.show()
```

# piechart

```python
plt.figure(figsize=(8,8))

df['category'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Distribution of News Categories")
plt.ylabel("")
plt.show()
```

# bar chart

```python
plt.figure(figsize=(10,5))

df['source'].value_counts().plot(kind='bar')

plt.title("Number of News Articles by Source")
plt.xlabel("News Source")
plt.ylabel("Number of Articles")
plt.xticks(rotation=45)
plt.show()
```

# boxplot

```python
df['text_length'] = df['text'].astype(str).apply(len)

plt.figure(figsize=(7,4))
sns.boxplot(y=df['text_length'])

plt.title("Distribution of News Text Length")
plt.ylabel("Text Length")
plt.show()
```
